# Nuages de mots

## Imports et stopwords

```python
from collections import Counter
from wordcloud import WordCloud
import os
import shutil
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from IPython.display import Image
```

```python
# Stopwords (Idem que dans s1)
sw = stopwords.words("french")
sw += ["les", "plus", "cette", "fait", "faire", "être", "deux", "comme", "dont", "tout",
       "ils", "bien", "sans", "peut", "tous", "après", "ainsi", "donc", "cet", "sous",
       "celle", "entre", "encore", "toutes", "pendant", "moins", "dire", "cela", "non",
       "faut", "trois", "aussi", "dit", "avoir", "doit", "contre", "depuis", "autres",
       "van", "het", "autre", "jusqu", "ville", "rossel", "dem"]

# Mes mots à ajouter à chaque essai (en minuscules)
EXTRA_STOPWORDS = ["demande", "bon", "bons", "bonne", "jeune", "fille", "maison", "louer",
                   "vendre", "agence", "rue", "belle", "gages"]

sw = set(sw) | set(EXTRA_STOPWORDS)
len(sw)
```

## Créer un fichier contenant le texte de tous les journaux d'une année donnée

```python
# Choisir une année
year = 1923

data_path = "../data"
if not os.path.exists(data_path):
    data_path = "../../data"
temp_path = os.path.join(data_path, "tmp")
temp_path
```

```python
# Imprimer le contenu du fichier de l'année et constater les "déchets"
with open(os.path.join(temp_path, f'{year}.txt'), 'r', encoding='utf-8') as f:
    before = f.read()

before[:500]
```

## Nettoyer le fichier à l'aide d'une fonction de nettoyage

### Créer la fonction de nettoyage (à adapter)

```python
def clean_text(year, folder=None):
    if folder is None:
        input_path = f"{year}.txt"
        output_path = f"{year}_clean.txt"
    else:
        input_path = f"{folder}/{year}.txt"
        output_path = f"{folder}/{year}_clean.txt"
    output = open(output_path, "w", encoding='utf-8')
    with open(input_path, 'r', encoding='utf-8') as f:
        text = f.read()
        words = nltk.wordpunct_tokenize(text)
        kept = [w.upper() for w in words if len(w) > 2 and w.isalpha() and w.lower() not in sw]
        kept_string = " ".join(kept)
        output.write(kept_string)
    return f'Output has been written in {output_path}!'
```

### Appliquer la fonction sur le fichier complet de l'année

```python
clean_text(year, folder=temp_path)
```

```python
# Vérifier le résultat
with open(os.path.join(temp_path, f'{year}_clean.txt'), 'r', encoding='utf-8') as f:
    after = f.read()

after[:500]
```

## Nuage de mots

```python
import re
corpus = open(os.path.join(data_path, "all.txt"), 'r', encoding='utf-8').read()
jours = "lundi|mardi|mercredi|jeudi|vendredi|samedi|dimanche"
mois = "janvier|f.vrier|mars|avril|mai|juin|juillet|ao.t|septembre|octobre|novembre|d.cembre"

def positions_des_dates(annee):
    motif = "(" + jours + r")\s+\d+\s+(" + mois + r")\s+" + str(annee)
    return [m.start() for m in re.finditer(motif, corpus, re.I)]

dates_1923 = positions_des_dates(1923)
milieu = dates_1923[len(dates_1923) // 2]
dates_1922 = [p for p in positions_des_dates(1922) if p < milieu]
dates_1924 = [p for p in positions_des_dates(1924) if p > milieu]
debut = min(p for p in dates_1923 if not dates_1922 or p > dates_1922[-1])
fin = dates_1924[0] if dates_1924 else len(corpus)
year_text = corpus[debut:fin]

os.makedirs(temp_path, exist_ok=True)
with open(os.path.join(temp_path, "1923.txt"), 'w', encoding='utf-8') as f:
    f.write(year_text)
print(len(year_text), "caractères écrits")
```

### Afficher les termes les plus fréquents

### Créer, stocker et afficher le nuage de mots

```python
cloud = WordCloud(width=2000, height=1000, background_color='white').generate_from_frequencies(frequencies)
cloud.to_file(os.path.join(temp_path, f"{year}.png"))
shutil.copy(os.path.join(temp_path, f"{year}.png"), f"nuage_{year}.png")
Image(filename=os.path.join(temp_path, f"{year}.png"))
```
